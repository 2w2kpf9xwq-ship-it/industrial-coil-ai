# 01 - Domain Analysis
## Industrial Coil Manufacturing Process

The goal of this project is to understand the manufacturing process of industrial electrical coils in order to develop a synthetic dataset for Machine Learning applications.
The dataset will not reproduce confidential production data but will simulate a realistic manufacturing process based on engineering assumptions.

## Company context
The company manufactures high quality electrical coils for industrial applications.

Main markets:
- Railway

Production characteristics:
- Small company
- High specialization
- Arisan production
- Few product families
- High quality standards

## Manufacturing process

1. Copper wire preparation
2. Wire tension control
3. Coil winding
4. Intermediate welding *(hypothesis)*
5. Electrical testing
6. PASS / FAIL
