# Notes

## Domain analysis: industrial coil manufacturing process

The goal of this project is to understand the manufacturing process of industrial electrical coils in order to develop a synthetic dataset for Machine Learning applications.
The dataset will not reproduce confidential production data but will simulate a realistic manufacturing process based on engineering assumptions.

## Company context

The company manufactures high quality electrical coils for industrial applications.

Main market:
- Railway

Production characteristics:
- Small company
- High specialization
- Artisan production
- Few product families
- High quality standards

## Manufacturing process

1. Copper wire preparation
2. Wire tension control
3. Coil winding
4. Intermediate welding *(hypothesis)*
5. Electrical testing
6. PASS / FAIL

## Process relationships and assumptions

### First relation: number of turns influences resistance

```text
Number of turns
    ↓
Coil winding
    ↓
Length of copper wire
    ↓
Resistance

Product characteristics
    |—— Wire material
    |—— Wire cross-section
    |—— Coil geometry
    |—— ...
        |
        ▼
    Process parameters
    |—— Number of turns
    |—— Speed
    |—— Tension
    |—— ...
        |
        ▼
    Coil produced
        |
        ▼
    Test measurements
```

### Second relation: temperature influences resistance

```text
Ambient temperature
    ↓
Coil / test conditions
    ↓
Resistance measurement

Welding temperature
    ↓
Quality of welded connection
    ↓
Electrical properties
```
